## Answers — Exercise 24.1

**(a)(i)** Two ways a device can be identified on a LAN:
1. **IP address** — a logical address assigned to the device on the network
2. **MAC address** — a unique hardware address permanently embedded in the network interface card

**(a)(ii)** Two reasons LANs need communication protocols:
1. To establish agreed rules for how data is formatted, transmitted and received, so that devices from different manufacturers can communicate with each other
2. To provide mechanisms for error detection and flow control, ensuring reliable and orderly transfer of data between devices

**(b)(i)** Packet-switching:
- Data is divided into smaller units called **packets**
- Each packet contains a **header** with source/destination IP address and a sequence number
- Packets travel **independently** through the network, possibly via different routes, and are **reassembled in the correct order** at the destination

**(b)(ii)** Disadvantage and how it is handled:
- **Disadvantage**: Packets may arrive **out of order** or with variable delays (jitter), which can disrupt time-sensitive applications
- **How handled**: Each packet carries a **sequence number** in its header, allowing the receiving device to buffer and reorder packets correctly before passing data to the application

**(b)(iii)** How a packet-switched network copes with a broken cable:
- Routers use routing protocols to **detect the failed link** and automatically **reroute packets** through an alternative path to reach the destination

**(c)** Actions carried out by the local DNS:
1. The local DNS **checks its own cache** to see if the IP address for the domain name is already stored
2. If not found, it queries a **root DNS server**, which directs it to the authoritative name server for that domain's TLD
3. The local DNS queries the **authoritative name server**, which returns the corresponding IP address
4. The local DNS **caches** the IP address for future requests and **returns it** to the requesting browser

**(d)(i)** **Encryption** — ensures only the intended recipient (who has the key) can read the message

**(d)(ii)** **Digital signature** — allows the recipient to verify the identity of the sender

## Answers — Exercise 24.2

**(a)(i)** Two advantages of packet switching over circuit switching:
1. **Better bandwidth utilisation** — multiple packets from different sources share the same link simultaneously, so bandwidth is not wasted when no data is being sent
2. **Greater resilience** — if one route fails, packets can be automatically rerouted via alternative paths without disrupting the communication

**(a)(ii)** One advantage of circuit switching over packet switching:
1. A **dedicated channel** guarantees consistent bandwidth and low, predictable latency — making it suitable for real-time voice calls where variable delay is unacceptable

**(b)(i)** One reason for using a parity check or checksum:
- To **detect errors** (bit corruption) that may have occurred during data transmission

**(b)(ii)** One example of an error a parity check cannot detect:
- If **two bits are flipped** simultaneously, the parity remains unchanged and the error goes undetected (even number of bit errors)

**(c)** Most significant difference between a switch and a router:
- A **switch** operates at the data link layer, using **MAC addresses** to forward frames between devices **within the same LAN**; a **router** operates at the network layer, using **IP addresses** to route packets **between different networks**

**(d)** Purpose of a bridge:
- A bridge **connects two network segments** and uses MAC addresses to selectively forward or filter frames, keeping local traffic within its own segment to reduce congestion

**(e)** Two advantages of client-server over peer-to-peer for storing shared data:
1. Data is stored **centrally on the server**, making it easier to manage, back up and maintain data consistency
2. The server can **enforce access control centrally**, ensuring only authorised users can read or modify shared data

## Exercise 24.5 2020/YIJC/P2/Q1 H2 Computing
In a Battleship game, a player fires missiles within a region measuring 6 metres by 6 metres, represented with full-stops ("`.`"), to sink the ships. Each ship is represented by "`XXX`".
The region is represented on the screen by a rectangular grid. Each square metre of the region is represented by an x-coordinate and a y-coordinate. The top left square metre of the region display has `x = 0` and `y = 0`.

### Task 1
In one of the games, three ships were positioned in the region as given in the text file `GAME.TXT`. Write a program to read in the data from this file, store it in a suitable data structure and display on the screen as shown.

    . . . . . .
    . . . . X .
    X X X . X .
    . . . . X .
    . X X X . .
    . . . . . .

<div style="text-align: right">[6]</div> 

````python
# Task 1 — Read GAME.TXT and display the region
# Assumes GAME.TXT has one "x,y" coordinate per line for each ship cell

def read_game_grid(filename):
    grid = [['.' for _ in range(6)] for _ in range(6)]
    with open(filename, 'r') as f:
        for line in f:
            line = line.strip()
            if line:
                parts = line.split(',')
                x, y = int(parts[0]), int(parts[1])
                grid[y][x] = 'X'
    return grid

def display_grid(grid):
    for row in grid:
        print(' '.join(row))

grid = read_game_grid('GAME.TXT')
display_grid(grid)
````

````python
# Task 2 — Fire 5 missiles; S = hit, O = miss, SSS = sunken ship

def find_ship_cells(actual_grid, y, x):
    """BFS to find all connected 'X' cells belonging to the same ship."""
    visited = set()
    queue = [(y, x)]
    ship_cells = []
    while queue:
        cy, cx = queue.pop(0)
        if (cy, cx) in visited:
            continue
        visited.add((cy, cx))
        if actual_grid[cy][cx] == 'X':
            ship_cells.append((cy, cx))
            for dy, dx in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
                ny, nx = cy + dy, cx + dx
                if 0 <= ny < 6 and 0 <= nx < 6 and (ny, nx) not in visited:
                    if actual_grid[ny][nx] == 'X':
                        queue.append((ny, nx))
    return ship_cells

def fire_missiles(actual_grid, num_missiles=5):
    hidden = [['.' for _ in range(6)] for _ in range(6)]

    for _ in range(num_missiles):
        display_grid(hidden)
        print()
        x = int(input("Enter x-coordinate: "))
        y = int(input("Enter y-coordinate: "))

        if actual_grid[y][x] == 'X':
            hidden[y][x] = 'S'
            # Check if the whole ship is now sunk
            ship_cells = find_ship_cells(actual_grid, y, x)
            if all(hidden[cy][cx] == 'S' for cy, cx in ship_cells):
                print("Ship sunk!")
        else:
            hidden[y][x] = 'O'
        print()

    # Reveal all remaining ship positions
    final = [row[:] for row in hidden]
    for r in range(6):
        for c in range(6):
            if actual_grid[r][c] == 'X' and final[r][c] == '.':
                final[r][c] = 'X'
    print("Final grid:")
    display_grid(final)

fire_missiles(grid)
````

### Task 3
Write server and client program for this asymmetric Battleship game where a server display the region and the player (the client) selects the target locations for his missiles. After firing each missile, the server returns an updated display for the region indicating a strike or a miss. Once the player fires his last missile, the server will display the positions of all the ships, together with the damages and misses, and both the client and server quit the game. <div style="text-align: right">[12]</div>

````python
# Task 3 — Client-Server Battleship
# Run SERVER code in one terminal, CLIENT code in another terminal.
# Uses socket programming (syllabus 4.1.6 — iterative server).

import socket

# ── HELPER ──────────────────────────────────────────────────────────────────
def grid_to_str(grid):
    return '\n'.join(' '.join(row) for row in grid)

def str_to_grid(s):
    return [row.split() for row in s.strip().split('\n')]

# ── SERVER ───────────────────────────────────────────────────────────────────
def battleship_server():
    actual_grid = read_game_grid('GAME.TXT')
    hidden = [['.' for _ in range(6)] for _ in range(6)]

    server_sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    server_sock.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
    server_sock.bind(('', 12345))
    server_sock.listen(1)
    print("Waiting for player (client)...")

    conn, addr = server_sock.accept()
    print(f"Player connected from {addr}")

    for _ in range(5):
        # Send hidden grid to client
        conn.send(grid_to_str(hidden).encode())
        conn.recv(4)  # wait for ACK

        # Receive target coordinates
        data = conn.recv(1024).decode().strip()
        x, y = int(data.split(',')[0]), int(data.split(',')[1])

        if actual_grid[y][x] == 'X':
            hidden[y][x] = 'S'
            ship_cells = find_ship_cells(actual_grid, y, x)
            if all(hidden[cy][cx] == 'S' for cy, cx in ship_cells):
                result = 'SUNK'
            else:
                result = 'HIT'
        else:
            hidden[y][x] = 'O'
            result = 'MISS'
        conn.send(result.encode())

    # Send final grid revealing all ships
    final = [row[:] for row in hidden]
    for r in range(6):
        for c in range(6):
            if actual_grid[r][c] == 'X' and final[r][c] == '.':
                final[r][c] = 'X'
    conn.send(grid_to_str(final).encode())
    print("Game over. Final grid sent.")
    conn.close()
    server_sock.close()

# ── CLIENT ───────────────────────────────────────────────────────────────────
def battleship_client():
    client_sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    client_sock.connect(('localhost', 12345))
    print("Connected to server.\n")

    for missile in range(1, 6):
        # Receive and display hidden grid
        raw = client_sock.recv(4096).decode()
        print(f"--- Missile {missile} ---")
        print(raw)
        client_sock.send(b'ACK ')  # signal ready for input

        x = int(input("Enter x-coordinate: "))
        y = int(input("Enter y-coordinate: "))
        client_sock.send(f"{x},{y}".encode())

        result = client_sock.recv(1024).decode()
        print(f"Result: {result}\n")

    # Receive and display final grid
    final_raw = client_sock.recv(4096).decode()
    print("=== Final grid ===")
    print(final_raw)
    client_sock.close()

# ── ENTRY POINT ──────────────────────────────────────────────────────────────
# Uncomment ONE of the lines below depending on which role this process plays:
# battleship_server()
# battleship_client()
````

## Task 4
The file `HANGMAN_WORDS.TXT` contains a list of words commonly found in an English dictionary. It contains words of varying length. You are to write a Python function named `getSecret()` to return the secret word for the player to guess. The function must include the following functionality:

- (a) Read the contents of the file `HANGMAN_WORDS.TXT` into a suitable data structure to facilitate the lookup of a word based on the length of the word.
- (b) Determine the shortest word and the longest word in the file.
- (c) Prompt the user to enter the length of the word to be guessed. The prompt must indicate the range of word length that the user can choose from based on the shortest and longest word found in Task 4(b). See example in the game play above.
    You **do not need** to write code to handle invalid input.
- (d) Look for a word from the data structure created in Task 4(a) that have the word length given by the user in Task 4(c). If there is more than one word with the word length, randomly choose one. For example, if the length of word chosen is 4. The code should randomly select one of the 4-letter word from the data structure created in Task 4(a).
- (e) Return the word from Task 4(d) for the player to guess. <div style="text-align: right">[6]</div>

````python
# Task 4 — getSecret(): load HANGMAN_WORDS.TXT, prompt for word length, return random word

import random
import math

def getSecret():
    word_dict = {}   # key: word length, value: list of words
    shortest = float('inf')
    longest = 0

    with open('HANGMAN_WORDS.TXT', 'r') as f:
        for line in f:
            word = line.strip().lower()
            if word:
                length = len(word)
                if length not in word_dict:
                    word_dict[length] = []
                word_dict[length].append(word)
                if length < shortest:
                    shortest = length
                if length > longest:
                    longest = length

    length_choice = int(input(f"Choose the length of word to guess:[{shortest} to {longest}]"))
    return random.choice(word_dict[length_choice])

secret = getSecret()
print(f"(Debug) Secret word: {secret}")
````

## Task 5
Write the code for a player to guess the secret word returned by the `getSecret()` function defined in Task 4. The number of guesses that a player can make is 2 times the length of the secret word to be guessed. For example, if the length of the secret word is 4, the number of attempts is 8.

The input prompt and output should be similar to the two game play examples shown above. There should be a number indicating the number of attempts left. 

You **do not need** to write code to handle invalid input. <div style="text-align: right">[7]</div>

````python
# Task 5 — Full Hangman game using getSecret()

def play_hangman():
    secret = getSecret()
    length = len(secret)
    attempts = 2 * length
    display = ['.' for _ in secret]

    print(f"You have {attempts} attempts. Good Luck!")

    while attempts > 0 and '.' in display:
        guess = input(f"[{attempts}]Guess letter:").lower()

        found = False
        for i, ch in enumerate(secret):
            if ch == guess and display[i] == '.':
                display[i] = ch
                found = True

        print(''.join(display))

        if not found:
            attempts -= 1

    if '.' not in display:
        print("WIN")
    else:
        print(f"LOSE..The word is {secret}")

play_hangman()
````

````python
# Task 7 — Hangman SERVER
# Server randomly selects word length and secret word; client guesses letters.
# Run this file first, then run the client (Task 8).

import socket, random, math

def load_word_dict():
    word_dict = {}
    with open('HANGMAN_WORDS.TXT', 'r') as f:
        for line in f:
            word = line.strip().lower()
            if word:
                length = len(word)
                if length not in word_dict:
                    word_dict[length] = []
                word_dict[length].append(word)
    return word_dict

def hangman_server():
    word_dict = load_word_dict()

    # Server randomly selects word length and secret word
    length = random.choice(list(word_dict.keys()))
    secret = random.choice(word_dict[length])
    attempts = 2 * length
    display = ['.' for _ in secret]

    # Reveal ceil(0.2 * length) letters
    num_reveal = math.ceil(0.2 * length)
    for pos in random.sample(range(length), num_reveal):
        display[pos] = secret[pos]

    server_sock = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    server_sock.setsockopt(socket.SOL_SOCKET, socket.SO_REUSEADDR, 1)
    server_sock.bind(('', 9998))
    server_sock.listen(1)
    print("Hangman server waiting for player...")

    conn, addr = server_sock.accept()
    print(f"Player connected from {addr}")

    # Send initial game state: "attempts,display_string"
    conn.send(f"{attempts},{''.join(display)}".encode())

    while attempts > 0 and '.' in display:
        guess = conn.recv(1024).decode().strip().lower()

        found = False
        for i, ch in enumerate(secret):
            if ch == guess and display[i] == '.':
                display[i] = ch
                found = True

        if not found:
            attempts -= 1

        if '.' not in display:
            status = 'WIN'
        elif attempts == 0:
            status = f'LOSE:{secret}'
        else:
            status = 'CONTINUE'

        conn.send(f"{attempts},{''.join(display)},{status}".encode())

    conn.close()
    server_sock.close()
    print("Game ended.")

hangman_server()
````

### Task 9
To prevent Taliban interference, messages sent between evacuees and the A330 MRTT are encrypted.

The encryption converts each character of the message into its ASCII number representation, adds the ASCII number representation of a character of a secret key which is in the same position, and then converts it back into ASCII text.

Secret keys are generated from passwords known only by the sender and receiver. Passwords that are shorter in length than the message are lengthened to match by repeating the password until the same length as the message is achieved.

For example, given that the inputs are

    password : "cat"
    message : "hello"

Step 1 - Extend password by repeating until matches length of message

    "cat" -> "catca"
    
Step 2 - Convert each character of password and message into ASCII number representation

    "hello" -> 104, 101, 108, 108, 111
    "catca" -> 99, 97, 116, 99, 97

Step 3 - Add the ASCII number representation of same positions together

    104+99, 101+97, 108+116, 108+99, 111+97 -> 203, 198, 224, 207, 208

Step 4 - Convert each number to its ASCII character

    203, 198, 224, 207, 208 -> "ËÆàÏÐ"

Therefore, the results are
    Secret key : "catca"
    Encrypted message : "ËÆàÏÐ"

Write a function `generateKey(password, message)` that takes in two argument strings message and password and returns the secret key.
Test your function using generateKey("cat", "hello")and show your output. <div style="text-align: right">[4]</div>

````python
# Task 9 — generateKey(password, message)
# Extends password by repeating until it matches the length of message.

def generateKey(password, message):
    key = ""
    for i in range(len(message)):
        key += password[i % len(password)]
    return key

# Test
print(generateKey("cat", "hello"))  # Expected output: catca
````

### Task 10
Write functions `encrypt(password, message)` and `decrypt(password, message)` that takes in `password` and message strings and returns the encrypted or decrypted `message`. It should use the above function `generateKey(password, message)` to obtain the secret key before performing the encryption.
Test your function with `decrypt("cat", encrypt("cat","hello"))`. Show your output.  <div style="text-align: right">[6]</div>

````python
# Task 10 — encrypt(password, message) and decrypt(password, message)
# Encryption: add ASCII values of message char + key char, convert back to char.
# Decryption: reverse — subtract key ASCII value.

def encrypt(password, message):
    key = generateKey(password, message)
    encrypted = ""
    for i in range(len(message)):
        encrypted += chr(ord(message[i]) + ord(key[i]))
    return encrypted

def decrypt(password, message):
    key = generateKey(password, message)
    decrypted = ""
    for i in range(len(message)):
        decrypted += chr(ord(message[i]) - ord(key[i]))
    return decrypted

# Test — decrypt should reverse encrypt exactly
result = decrypt("cat", encrypt("cat", "hello"))
print(result)  # Expected output: hello
````

### Task 11
Since the A330 MRTT plane can only take a maximum of 266 passengers at once, implement a Queue to manage incoming pickup requests and dropoffs. Write a class `Queue` with the following
methods:
- `enqueue(string)` which takes in a string and adds it to the queue or returns `"Queue is full!"` if queue is full
- `dequeue` which removes the next item for the queue or returns `"Queue is empty!"` if  the queue is empty
- `display` which returns a string of queue items in sequence from head to tail or returns `"Queue is empty!"` if the queue is empty
- `current_size` which returns an integer of the number of items in the queue. <div style="text-align: right">[8]</div>

````python
# Task 11 — Queue class (max capacity 266 passengers)
# enqueue(string), dequeue(n), display(), current_size()

class Queue:
    def __init__(self, max_size=266):
        self.__data = []
        self.__max_size = max_size

    def enqueue(self, string):
        if len(self.__data) >= self.__max_size:
            return "Queue is full!"
        self.__data.append(string)

    def dequeue(self, n=1):
        if not self.__data:
            return "Queue is empty!"
        for _ in range(min(n, len(self.__data))):
            self.__data.pop(0)

    def display(self):
        if not self.__data:
            return "Queue is empty!"
        lines = []
        for i in range(0, len(self.__data), 10):
            chunk = self.__data[i:i + 10]
            lines.append(", ".join(chunk) + ",")
        # Remove trailing comma from the very last item
        last = lines[-1].rstrip(",")
        lines[-1] = last
        return "\n".join(lines)

    def current_size(self):
        return len(self.__data)
````
