Your program code for the stack.    
<div style="text-align: right">[10]</div>

````python
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if self.is_empty():
            raise IndexError("Empty stack cuh")
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            raise IndexError("Empty stack cuh")
        return self.items[-1]

    def size(self):
        return len(self.items)
````

## Task 1

Write the Linkedlist class in Python. Use of a simple Python list is not sufficient. Include the following methods:
- `insert(integer_value)` inserts the `integer_value` at the beginning (head) of the list
- `delete(integer_value)` attempts to delete `integer_value` from the list; if the item was not present, return `None`
- `search(integer_value)` returns a Boolean value: `True` if `integer_value` is in the list, `False` if not in the list
- `count()` should return the number of elements in the list, or zero if empty
- `to_String(i)` should return a string containing a suitably formatted list with the elements separated by a comma and a space, with square brackets at either end, e.g. in the form: 
    `[11, 2, 7, 4]`

<div style="text-align: right">[8]</div> 

Test `LinkedList` by using the data in the file `Task3data.txt`. Use the `to_String()` method to print the resulting contents of the list.

<div style="text-align: right">[3]</div>

````python
class Linkedlist:
    def __init__(self, value):
        self.value = value
        self.next = None

    def insert(self, integer_value):
        if self.search(integer_value):
            return self

        new_node = Linkedlist(integer_value)
        new_node.next = self
        return new_node

    def delete(self, integer_value):
        if self.value == integer_value:
            return self.next

        current = self
        while current.next is not None:
            if current.next.value == integer_value:
                current.next = current.next.next
                return self
            current = current.next

        return self

    def search(self, integer_value):
        current = self
        while current is not None:
            if current.value == integer_value:
                return True
            current = current.next
        return False

    def count(self):
        total = 0
        current = self
        while current is not None:
            total += 1
            current = current.next
        return total

    def to_String(self):
        elements = ""
        current = self
        while current is not None:
            elements += str(current.value)
            if current.next is not None:
                elements += ", "
            current = current.next
        return "[" + elements + "]"


with open('./resources/Task3data.txt') as f:
    data = [int(line.strip()) for line in f]

head = Linkedlist(data[0])
for record in data:
        head = head.insert(record)
print(head.to_String())
````

````python
# Basic Binary Search Tree Code
class TreeNode:
    def __init__(self, key):
        # Stores the key (value) of the node
        self.key = key
        
        # Reference to the left child
        self.left = None
        
        # Reference to the right child
        self.right = None


class BinarySearchTree:
    def __init__(self):
        # Reference to the root node of the BST
        self.root = None

    def insert(self, key):
        # Public method to insert a key into the BST
        self.root = self.insert_helper(self.root, key)

    def insert_helper(self, root, key):
        # Recursively inserts a key into the correct position in the BST
        if root is None:
            # Create a new node if we reached an empty position
            return TreeNode(key)

        # Recur down the left subtree if key is smaller
        if key < root.key:
            root.left = self.insert_helper(root.left, key)
        
        # Recur down the right subtree if key is larger
        elif key > root.key:
            root.right = self.insert_helper(root.right, key)

        # Return the (possibly unchanged) root node
        return root

    def search(self, key):
        # Public method to search for a key in the BST
        return self.search_helper(self.root, key)

    def search_helper(self, root, key):
        # Recursively searches for a key starting from a given node
        if root is None or root.key == key:
            # Return the node if found, or None if not found
            return root

        # Search left subtree if key is smaller
        if key < root.key:
            return self.search_helper(root.left, key)
        else:
            # Search right subtree if key is larger
            return self.search_helper(root.right, key)

    def inorder_traversal(self):
        # Returns a list of keys using inorder traversal (Left → Root → Right)
        result = []
        self.inorder_helper(self.root, result)
        return result

    def inorder_helper(self, root, result):
        # Recursively performs inorder traversal
        if root:
            self.inorder_helper(root.left, result)
            result.append(root.key)
            self.inorder_helper(root.right, result)

    def preorder_traversal(self):
        # Returns a list of keys using preorder traversal (Root → Left → Right)
        result = []
        self.preorder_helper(self.root, result)
        return result

    def preorder_helper(self, root, result):
        # Recursively performs preorder traversal
        if root:
            result.append(root.key)
            self.preorder_helper(root.left, result)
            self.preorder_helper(root.right, result)
    
    def postorder_traversal(self):
        # Returns a list of keys using postorder traversal (Left → Right → Root)
        result = []
        self.postorder_helper(self.root, result)
        return result

    def postorder_helper(self, root, result):
        # Recursively performs postorder traversal
        if root:
            self.postorder_helper(root.left, result)
            self.postorder_helper(root.right, result)
            result.append(root.key)
    
tree = BinarySearchTree()
with open('./resources/FRUITS.txt') as f:
    fruits = [x.strip() for x in f]
    print(fruits)

for fruit in fruits: 
    tree.insert(fruit)

tree.preorder_traversal()
````
